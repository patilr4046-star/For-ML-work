# Data Preprocessing & Cleaning

```python
df_encode = pd.get_dummies(df, drop_first = True )
```

```python
df_encode
```

```python
df_encode['Oldpeak_Category'] = pd.cut(
    df_encode['Oldpeak'],
    bins = [-np.inf, 0, 1, 2, np.inf],
    labels = ['Zero_or_Lower', 'Mild', 'Moderate', 'Higher'],
    right=False
)
```

```python
df_encode = pd.get_dummies(df_encode, columns = ['Oldpeak_Category'])
```

```python
df_encode = df_encode.astype(int)
```

```python
df_encode
```

```python
from sklearn.preprocessing import StandardScaler
cols = ['Age','RestingBP','Cholesterol','MaxHR','Oldpeak']
scaler = StandardScaler()
df_encode[cols]= scaler.fit_transform(df_encode[cols])
df_encode.head()
```

```python
from scipy.stats import pearsonr

# Pearson Correlation Calculation

selected_features = [
    'Age', 'RestingBP', 'Cholesterol', 'FastingBS', 'MaxHR', 'Oldpeak',
    'Sex_M', 'ChestPainType_ATA', 'ChestPainType_NAP',
    'ChestPainType_TA', 'RestingECG_Normal', 'RestingECG_ST',
    'ExerciseAngina_Y', 'ST_Slope_Flat', 'ST_Slope_Up',
       'Oldpeak_Category_Zero_or_Lower', 'Oldpeak_Category_Mild',
       'Oldpeak_Category_Moderate', 'Oldpeak_Category_Higher'
]



correlations = {
    feature: pearsonr(df_encode[feature],df_encode['HeartDisease'])[0]
    for feature in selected_features
}

correlation_df = pd.DataFrame(list(correlations.items()),columns=['Feature', 'Pearson Correlation'])
correlation_df = correlation_df.sort_values(by='Pearson Correlation',ascending=False)
```

```python
correlation_df
```

```python
cat_features = [
    'Age', 'RestingBP', 'Cholesterol', 'FastingBS', 'MaxHR', 'Oldpeak',
       ', 'Sex_M', 'ChestPainType_ATA', 'ChestPainType_NAP',
       'ChestPainType_TA', 'RestingECG_Normal', 'RestingECG_ST',
       'ExerciseAngina_Y', 'ST_Slope_Flat', 'ST_Slope_Up',
       'Oldpeak_Category_Zero_or_Lower', 'Oldpeak_Category_Mild',
       'Oldpeak_Category_Moderate', 'Oldpeak_Category_Higher'
]
```

```python

from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(
        df_encode[col],
        df_encode['HeartDisease']
    )

    chi2_stat, p_val, _, _ = chi2_contingency(contingency)

    decision = (
        'Reject Null (Keep Feature)'
        if p_val < alpha
        else 'Fail to Reject Null (Review Feature)'
    )

    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')

chi2_df
```
